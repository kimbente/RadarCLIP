# Trace plots

We collapse the x-axis to show the amplitude curved of each column. 

```python
def plot_traces_from_tile(tile_tensor, i):
    # Extract tile data for the specified index
    tile = tile_tensor[i, 0, :, :].numpy()

    n_rows = tile.shape[0]
    rows = np.arange(n_rows)

    fig, ax = plt.subplots(figsize = (8, 8))
    for col in range(tile.shape[-1]):
        ax.plot(tile[:, col], rows, alpha = 0.2, color = 'black', linewidth = 0.2)

    # cut off y-axis at n_rows to match the image and invert
    ax.set_ylim(n_rows, 0)
    ax.set_xlabel("Amplitude")
    ax.set_ylabel("Sample index")
    ax.set_title(f"Traces for tile {i}")

    ax.set_box_aspect(1)
    plt.show()
```

```python
plot_traces_from_tile(tile_tensor, 280)
```
