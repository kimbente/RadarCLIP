# Download

This notebook downloads all granules (i.e, radargrams) from both [IceBridge HiCARS 1 L1B Time-Tagged Echo Strength Profiles V001](https://www.earthdata.nasa.gov/data/catalog/nsidc-cprd-ir1hi1b-1) and [IceBridge HiCARS 2 L1B Time-Tagged Echo Strength Profiles V001](https://www.earthdata.nasa.gov/data/catalog/nsidc-cprd-ir2hi1b-1) within the Australian Antarctic Territory. 

We use [earthaccess](https://earthaccess.readthedocs.io/en/stable/), the python library for NASA Easrthdata access to download these granules. 

## Login

First, we need to authenticate/login.

```python
import os
os.environ["EARTHDATA_USERNAME"] = "insertyourusernamehere"
os.environ["EARTHDATA_PASSWORD"] = "insertyourpasswordhere"

import earthaccess
auth = earthaccess.login(strategy = "environment", persist = True)
# Alternative strategy (did not work)
# auth = earthaccess.login(strategy = "interactive", persist = True)
print(auth.authenticated)
```

## Download HiCARS1 (IR1HI1B)

1152 granules (101 GB)

```python
# search for data
results_HiCARS_1 = earthaccess.search_data(
    short_name = 'IR1HI1B',
    # (west, south, east, north)
    # Australian boundries
    bounding_box = (45, -90, 160, -50)
)
# returns granules
print(len(results_HiCARS_1))
print(results_HiCARS_1)
```

```python
if redownload_radargrams:
# Takes around an hour
    files = earthaccess.download(results_HiCARS_1, path_to_IR1HI1B)
```
